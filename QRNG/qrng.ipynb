```python
"""
Example to demonstrate how a QRNG works

"""

from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator, Aer

# Number of bits in string of random number
n = 6

q = QuantumRegister(n, "q")
c = ClassicalRegister(n, "measurement")

qc = QuantumCircuit(q, c)
#print(qc.qubits[0])

# Create superpositions using Hadamards on all n qubits
for i in range(n):
    qc.h(q[i])
    qc.measure(q[i], c[i])

# Visualize circuit
qc.draw('mpl')

```

```python
# Continuation
simulator = AerSimulator()

qct = transpile(qc, simulator)
counts = simulator.run(qct, shots=1).result().get_counts()

b_num = list(counts.keys())[0]
# Print the random number as a binary string
print("Random number (binary representation): ", b_num)

# Convert to decimal and print
decimal = int(b_num, 2)
print("\nRandom number in decimal: ", decimal)
```

```python
import math

def num_bits(n):
    return (math.floor(math.log(n, 2)) + 1)
```

```python
# Function to return a single random number
def rng(n_bits=8):
    
    """
    Generate random numbers using quantum superposition (H gates).
    
    Args:
        n_bits (int): Number of random bits to generate
    
    Returns:
        int: Random integer value
        list: List of measurement outcomes (0s and 1s)
    """
    
    # Create a QC with n_bits q- and c- bits
    qc = QuantumCircuit(n_bits, n_bits)

    # Apply Hadamard gate to each qubit
    for q_i in range(n_bits):
        qc.h(q_i)

    # Append a measurement gate for each qubit
    for i in range(n_bits):
        qc.measure(i, i)

    # Use qiskit Aer simulator
    simulator = AerSimulator()

    # Execute circuit
    qct = transpile(qc, simulator)
    counts = simulator.run(qct, shots=1).result().get_counts()

    # Store state of qubits as a binary string
    binary_string = list(counts.keys())[0]

    return int(binary_string, 2), [int(bit) for bit in binary_string]


# Function that returns multiple random numbers based on rng defined above
def multiple_rng(n_num, n_bits=8):
    """
    Generate multiple quantum random numbers.
    
    Args:
        n_num (int): How many random numbers to generate
        n_bits (int): Number of bits per random number
    
    Returns:
        list: List of random integers
    """
    
    decimal = []
    binary = []
    for i in range(n_num):
        num = rng(n_bits)
        decimal.append(num[0])
        binary.append(num[1])

    return decimal, binary


# Test the above functions
if __name__ == "__main__":

    # Input number of bits
    n = 8
    
    # Get a random number (decimal plus binary representation)
    r_num = rng(n)

    # Print the output
    print(f"SINGLE RANDOM NUMBER GENERATOR ({n} bits)")
    print(f"\nRandom number (in decimal): {r_num[0]} ")
    
    #print(f"\nBinary representation: {r_num[1]}")

    print(f"\nBinary string: {''.join(map(str, r_num[1]))}")
    print("-"*50)

    # Generate multiple random numbers
    
    num_numbers = 5
    num_bits = 5

    """
    # Input parameter values
    int(input("n (number of bits in number): "))

    int(input("no. of numbers to generate: "))

    int(input("Number of bits: "))
    """

    # Generate and print random numbers
    r_nums, bit_reps = multiple_rng(num_numbers, num_bits)
    print(f"\nMULTIPLE RANDOM NUMBER GENERATOR ({num_bits} bits)")
    print(f"\nRandom numbers: {r_nums}")

    # Pring bit strings for numbers
    bit_strings = []
    print("\nBinary strings: ", end='')
    for i in range(len(bit_reps)):
        bit_strings.append(''.join(map(str, bit_reps[i])))
    
    print(bit_strings)
```
